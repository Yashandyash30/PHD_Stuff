```python
for i in range(1,6):
    n5_pha= "bn140304557_n5_srcspectra.pha{%d}"%(i)
    n5_rsp= "bn140304557_n5_weightedrsp.rsp{%d}"%(i)
    n5_bak= "bn140304557_n5_bkgspectra.bak{%d}"%(i)
    
    nain5=OGIPLike("GBM_n5", observation= n5_pha, response= n5_rsp, background = n5_bak  )
    nain5.set_active_measurements("8-33", "37-900")
    nain5.display()
```

```python
fit_data = np.zeros(20)


for i in range(1,6):
    n3_pha= "bn140304557_n3_srcspectra.pha{%d}"%(i)
    n3_rsp= "bn140304557_n3_weightedrsp.rsp{%d}"%(i)
    n3_bak= "bn140304557_n3_bkgspectra.bak{%d}"%(i)
    

    nain3=OGIPLike("GBM_n3", observation= n3_pha, response= n3_rsp, background = n3_bak  )
    nain3.set_active_measurements("8-33", "37-900")
    #######################################################
    
    n4_pha= "bn140304557_n4_srcspectra.pha{%d}"%(i)
    n4_rsp= "bn140304557_n4_weightedrsp.rsp{%d}"%(i)
    n4_bak= "bn140304557_n4_bkgspectra.bak{%d}"%(i)
    

    nain4=OGIPLike("GBM_n4", observation= n4_pha, response= n4_rsp, background = n4_bak  )
    nain4.set_active_measurements("8-33", "37-900")
    #######################################################
    
    n5_pha= "bn140304557_n5_srcspectra.pha{%d}"%(i)
    n5_rsp= "bn140304557_n5_weightedrsp.rsp{%d}"%(i)
    n5_bak= "bn140304557_n5_bkgspectra.bak{%d}"%(i)
    

    nain5=OGIPLike("GBM_n5", observation= n5_pha, response= n5_rsp, background = n5_bak  )
    nain5.set_active_measurements("8-33", "37-900")
    #######################################################
    
    b0_pha= "bn140304557_b0_srcspectra.pha{%d}"%(i)
    b0_rsp= "bn140304557_b0_weightedrsp.rsp{%d}"%(i)
    b0_bak= "bn140304557_b0_bkgspectra.bak{%d}"%(i)

    bgob0=OGIPLike("GBM_b0", observation= b0_pha, response= b0_rsp, background = b0_bak)
    bgob0.set_active_measurements("200-40000")

    
    band = Cutoff_powerlaw()+Blackbody()
    model = Model(PointSource("GBM", RA, DEC, spectral_shape=band)) #GRB = PointSource( triggerName, ra, dec, band )
    model.GBM.spectrum.main.composite.K_1.prior= Log_uniform_prior(lower_bound=1e-6, upper_bound=1e1)
    model.GBM.spectrum.main.composite.index_1.prior= Truncated_gaussian(lower_bound=-2, upper_bound=1, mu=-1, sigma=0.5)
    model.GBM.spectrum.main.composite.xc_1.prior= Truncated_gaussian(lower_bound=10, upper_bound=1000, mu=300, sigma=200)
    model.GBM.spectrum.main.composite.K_2.prior=Log_uniform_prior(lower_bound=1e-9, upper_bound=1e-2)
    model.GBM.spectrum.main.composite.kT_2.prior = Truncated_gaussian(lower_bound=0.1, upper_bound=50,mu=10,sigma=7)

    
    data_list = DataList(nain3,nain4,nain5,bgob0)
    bayes_analysis = BayesianAnalysis(model, data_list)

    bayes_analysis.set_sampler("multinest")
    bayes_analysis.sampler.setup(n_live_points=10000)
    bayes_analysis.sample()
    fig = plot_spectra(bayes_analysis.results,ene_min=1, ene_max=10e4, num_ene=200,contour_colors='white',contour_style_kwargs=dict(alpha=0.2), use_components= True, components_to_use=['total'],flux_unit='keV2/(cm2 s keV)')
    plt.xlabel('Energy (keV)',fontsize='20')
    plt.ylabel(r'$\rm \nu$ $\rm F_{\rm \nu}$ $\rm (keV$ $\rm cm^{-2}$ $\rm sec^{-1})$',fontsize='20')
    fig = bayes_analysis.results.corner_plot()
    fig.savefig('cut_bb_spec_%d'%(i))
    a=bayes_analysis.results.get_point_source_flux(10 * u.keV, 10 * u.MeV)
    print(a)
    
    table1 = bayes_analysis.results.get_data_frame("equal tail")
    table1 = table1.drop('unit', axis=1)
    table1 = table1.to_numpy().reshape((20))
    
    fit_data = np.vstack((fit_data, table1))
```

```python
fit_data
cutbb_fit_param = Table(fit_data[1:], names=['B_K', 'B_K_p_err', 'B_K_n_err', 'B_K_avg_err', 'B_alpha', 'B_alpha_p_err', 'B_alpha_n_err', 'B_alpha_avg_err', 'E_c', 'E_c_p_err', 'E_c_n_err', 'E_c_avg_err','Bb_K', 'Bb_K_p_err', 'Bb_K_n_err', 'Bb_K_avg_err','Bb_Kt', 'Bb_Kt_p_err', 'Bb_Kt_n_err', 'Bb_Kt_avg_err'])
cutbb_fit_param.write('cutoffpl_bb.csv', overwrite = True)
cutbb_fit_param
```
